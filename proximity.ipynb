```python
import pandas as pd
```

```python
bigfoot_df = pd.read_csv("data/combined_bigfoot_v1.csv")
uap_df = pd.read_csv("data/us_uap_1940_v1.csv")
```

```python
proximity_df = pd.merge(
    bigfoot_df[[
        'bf_id', 'latitude', 'longitude', 'full_date',
        'geohash_7', 'geohash_6', 'geohash_5', 
        'state_code', 'year'
    ]],
    uap_df[[
        'uap_id', 'full_date', 'city', 'state_code', 
        'shape_group', 'geohash_7', 'geohash_6', 'geohash_5', 'year'
    ]],
    on='geohash_6',        
    how='inner'
)

proximity_df.shape
```
